```python
import numpy as np
import sounddevice as sd
import matplotlib.pyplot as plt
from scipy.signal import square
from scipy.signal import sawtooth

# Nhập thông số từ bàn phím
A = 500

# Thông số cố định
f = 100      # Hz
phi = 0       # rad
fs = 100*f    # Hz
duration = 2  # giây

# Tạo trục thời gian
t = np.arange(0, duration, 1/fs)

# Tạo sóng sin
#x = A * np.sin(2 * np.pi * f * t + phi)
x = A * sawtooth(2 * np.pi * f * t + phi, width=0.5)
#x = A * square(2 * np.pi * f * t + phi)

# Phát âm thanh
print("Đang phát âm thanh...")
sd.play(x, fs)
sd.wait()

print("Đã phát xong!")

plt.figure(figsize=(10, 4))

# Chỉ vẽ 10 ms đầu tiên để nhìn rõ dạng sóng
N = int(0.01 * fs)

plt.plot(t[:N], x[:N])

plt.title("Xung vuông 440 Hz")
plt.xlabel("Thời gian (s)")
plt.ylabel("Biên độ")
plt.grid()

plt.show()
```
