```python
#Viết hàm chorus_stereo(x, fs) mô phỏng dàn hòa tấu 3 người chơi:
import numpy as np
from scipy.io.wavfile import read, write
from IPython.display import Audio, display


def chorus_stereo(x, fs):

    D0_L_ms = 15.0
    D0_R_ms = 20.0

    FLFO1 = 0.8
    FLFO2 = 1.1

    theta1 = 0
    theta2 = np.pi / 2

    delta_D_ms = 5.0

    D0_L = D0_L_ms * 1e-3 * fs
    D0_R = D0_R_ms * 1e-3 * fs
    delta_D = delta_D_ms * 1e-3 * fs

    N = len(x)

    n = np.arange(N)

    D_L = D0_L + delta_D * np.sin(
        2 * np.pi * FLFO1 * n / fs + theta1
    )

    D_R = D0_R + delta_D * np.sin(
        2 * np.pi * FLFO2 * n / fs + theta2
    )

    left = np.zeros(N)
    right = np.zeros(N)


    direct_gain = 0.5

    for i in range(N):

     # LEFT

        D_int = int(np.floor(D_L[i]))
        delta = D_L[i] - D_int

        if i - D_int - 1 >= 0:

            delayed_L = (
                (1 - delta) * x[i - D_int]
                + delta * x[i - D_int - 1]
            )

        elif i - D_int >= 0:

            delayed_L = x[i - D_int]

        else:

            delayed_L = 0


        # RIGHT

        D_int = int(np.floor(D_R[i]))
        delta = D_R[i] - D_int

        if i - D_int - 1 >= 0:

            delayed_R = (
                (1 - delta) * x[i - D_int]
                + delta * x[i - D_int - 1]
            )

        elif i - D_int >= 0:

            delayed_R = x[i - D_int]

        else:

            delayed_R = 0


        left[i] = direct_gain * x[i] + 0.5 * delayed_L

        right[i] = direct_gain * x[i] + 0.5 * delayed_R


    stereo = np.column_stack((left, right))

    return stereo
```

```python
import numpy as np
from scipy.io.wavfile import read, write
from IPython.display import Audio, display

fs, x = read("guitar1.wav")
x = x.astype(float)
if x.ndim == 2:
    x = np.mean(x, axis=1)

x = x / np.max(np.abs(x))

y_stereo = chorus_stereo(x, fs)

print("Kích thước:", y_stereo.shape)

y_stereo = y_stereo / np.max(np.abs(y_stereo))

y_int16 = np.int16(y_stereo * 32767)

file_name = "guitar_chorus_stereo.wav"
write(file_name, fs, y_int16)

print("Đã xuất:", file_name)


display(Audio(filename=file_name))
```
